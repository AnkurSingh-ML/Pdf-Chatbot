# Character Splitting

* https://chunkviz.up.railway.app/

It is the process of simply dividing your text into N-character sized chunks regardsless of their content of form.

* **chunk_size** : The number of characters upu would like in your chunks 50, 100, 120 etc.
* **Chunk Overlap** : The amount you would like your sequential chunka to overlap.

```python
text = "This is the text I would like to chunk up. It is the example text for this exercise"
print(text)
```

```python
# create a list that will hold the chunks
chunks = []

chunk_size = 33 # no of characters in each chunk

# Run through a range with the length of your text and iterate every chunk_size you want
for i in range(0, len(text), chunk_size):
    chunk = text[i: i+ chunk_size]
    chunks.append(chunk)
chunks
```

```python
from langchain.text_splitter import CharacterTextSplitter
text_splitter = CharacterTextSplitter(chunk_size=35, chunk_overlap=0, separator='', strip_whitespace=False)
```

```python
text_splitter.create_documents([text])
```

## Chunk Overlap and Separators

Chunk overlap will blend together our chunks so that the tail of Chunk #1 will be the same thing and the head of Chunk #2 and so on ans so forth.

```python
text_splitter = CharacterTextSplitter(chunk_size=35, chunk_overlap = 4, separator = '')
```

```python
text_splitter = CharacterTextSplitter(chunk_size=35, chunk_overlap = 0, separator = 'ch')
text_splitter.create_documents([text])
```
